```python
import nltk
import unicodedata
from collections import Counter

nltk.download("indian", quiet=True)

from nltk.corpus import indian
from nltk.tokenize import wordpunct_tokenize

print("NLTK version:", nltk.__version__)
print("Available Indian corpus files:", indian.fileids())
```

```python
# Load sentences from the NLTK Indian corpus.
all_sentences = []

for fileid in indian.fileids():
    for sentence in indian.sents(fileids=fileid):
        words = []
        for token in sentence:
            words.append(token[0] if isinstance(token, tuple) else str(token))
        all_sentences.append(words)

print("Total sentences available:", len(all_sentences))
```

```python
# Select the first 500 sentences.
sentences = all_sentences[:500]

print("Selected sentences:", len(sentences))

for i, sentence in enumerate(sentences[:5], 1):
    print(f"{i}.", " ".join(sentence))
```

```python
# Unicode-aware tokenization and basic preprocessing.
text = " ".join(" ".join(sentence) for sentence in sentences)
text = unicodedata.normalize("NFC", text)

tokens = wordpunct_tokenize(text)
tokens = [token for token in tokens if any(char.isalnum() for char in token)]

print("First 50 tokens:")
print(tokens[:50])
```

```python
total_tokens = len(tokens)
vocabulary_size = len(set(tokens))

print("Total tokens:", total_tokens)
print("Vocabulary size:", vocabulary_size)
```

```python
top_10 = Counter(tokens).most_common(10)

print("Top 10 most frequent tokens:")
for token, count in top_10:
    print(f"{token!r}: {count}")
```

```python
example = " ".join(sentences[0])

print("Original sentence:")
print(example)

print("\nTokens:")
print(wordpunct_tokenize(example))
```
