```python
import re
import unicodedata
import pandas as pd
```

```python
# At least 20 noisy Indic-language samples.
noisy_samples = [
    "नमस्ते   दुनिया!!!",
    "यह   एक   परीक्षण है....",
    "आज मौसम बहुत अच्छा है!!!",
    "आप कैसे हैं???",
    "मुझे NLP बहुत पसंद है!!!",
    "दिल्ली में   आज ३०°C है.",
    "मेरा नाम Rahul है!!!",
    "कृपया   जल्दी   आइए......",
    "यह बहुत अच्छा है!!!!!!",
    "कक्षा २०२६ में शुरू होगी.",
    "आपका स्वागत है   !!!",
    "मुझे Machine LEARNING पसंद है.",
    "क्या आप मेरी मदद कर सकते हैं???",
    "यह संदेश   बहुत    महत्वपूर्ण है!!!",
    "आज १२ बजे meeting है.",
    "बहुत बढ़िया!!!!!! काम किया.",
    "कृपया अपना नाम   बताइए...",
    "मुझे Python सीखना है!!!",
    "यह   data बहुत useful है.",
    "धन्यवाद   :)   धन्यवाद"
]

print("Number of samples:", len(noisy_samples))
```

```python
def normalize_digits(text):
    result = []
    for char in text:
        if char.isdecimal():
            result.append(str(unicodedata.digit(char)))
        else:
            result.append(char)
    return "".join(result)
```

```python
def normalize_text(text):
    text = unicodedata.normalize("NFC", text)

    punctuation_map = str.maketrans({
        "“": '"', "”": '"', "‘": "'", "’": "'",
        "–": "-", "—": "-", "…": "...",
        "。": ".", "，": ",", "！": "!", "？": "?"
    })
    text = text.translate(punctuation_map)

    text = normalize_digits(text)
    text = re.sub(r"([!?.])\1+", r"\1", text)
    text = re.sub(r"\s+", " ", text).strip()

    # Case normalization applies to Roman-script words.
    text = text.lower()

    return text
```

```python
normalized_samples = [normalize_text(text) for text in noisy_samples]

df = pd.DataFrame({
    "Original Text": noisy_samples,
    "Normalized Text": normalized_samples
})

pd.set_option("display.max_colwidth", None)
df
```

## Observations

Normalization reduces superficial variation before downstream NLP tasks. It can make tokenization, vocabulary statistics, matching, retrieval, and machine-learning inputs more consistent.

The original text is preserved alongside normalized text so the source remains available.
