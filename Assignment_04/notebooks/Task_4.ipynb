```python
import re
import unicodedata
import numpy as np
import pandas as pd
import nltk

nltk.download("punkt", quiet=True)
from nltk.tokenize import wordpunct_tokenize
```

```python
parallel_pairs = [
    ("मैं आज कॉलेज जा रहा हूँ।", "I am going to college today."),
    ("मेरा नाम राहुल है।", "My name is Rahul."),
    ("आज मौसम बहुत अच्छा है।", "The weather is very nice today."),
    ("मुझे किताबें पढ़ना पसंद है।", "I like reading books."),
    ("वह रोज़ सुबह दौड़ता है।", "He runs every morning."),
    ("हमने कल एक फिल्म देखी।", "We watched a movie yesterday."),
    ("कृपया दरवाज़ा बंद कर दीजिए।", "Please close the door."),
    ("भारत एक बड़ा देश है।", "India is a large country."),
    ("मुझे कंप्यूटर विज्ञान पसंद है।", "I like computer science."),
    ("वह अपनी परीक्षा की तैयारी कर रही है।", "She is preparing for her exam."),
    ("आज मेरी कक्षा जल्दी खत्म होगी।", "My class will finish early today."),
    ("हम रविवार को बाजार जाएंगे।", "We will go to the market on Sunday."),
    ("यह किताब बहुत उपयोगी है।", "This book is very useful."),
    ("मेरे दोस्त दिल्ली में रहते हैं।", "My friends live in Delhi."),
    ("मैं रोज़ Python सीखता हूँ।", "I learn Python every day."),
    ("कृपया मुझे पानी दीजिए।", "Please give me some water."),
    ("बच्चे पार्क में खेल रहे हैं।", "The children are playing in the park."),
    ("आज हमें एक प्रोजेक्ट पूरा करना है।", "We have to complete a project today."),
    ("वह ट्रेन से घर गया।", "He went home by train."),
    ("अच्छी नींद स्वास्थ्य के लिए जरूरी है।", "Good sleep is important for health.")
]

parallel_df = pd.DataFrame(parallel_pairs, columns=["Hindi", "English"])
parallel_df.index = range(1, len(parallel_df) + 1)
parallel_df.index.name = "Pair ID"

parallel_df
```

```python
def tokenize_text(text):
    text = unicodedata.normalize("NFC", text)
    tokens = wordpunct_tokenize(text)
    return [t for t in tokens if any(ch.isalnum() for ch in t)]

parallel_df["Hindi Tokens"] = parallel_df["Hindi"].apply(tokenize_text)
parallel_df["English Tokens"] = parallel_df["English"].apply(tokenize_text)
parallel_df["Hindi Length"] = parallel_df["Hindi Tokens"].apply(len)
parallel_df["English Length"] = parallel_df["English Tokens"].apply(len)

parallel_df
```

```python
# Basic one-to-one sentence alignment.
for pair_id, row in parallel_df.iterrows():
    print(f"Pair {pair_id}")
    print("Hindi  :", row["Hindi"])
    print("English:", row["English"])
    print()
```

```python
hindi_tokens = [t for row in parallel_df["Hindi Tokens"] for t in row]
english_tokens = [t for row in parallel_df["English Tokens"] for t in row]

statistics = pd.DataFrame({
    "Language": ["Hindi", "English"],
    "Total Tokens": [len(hindi_tokens), len(english_tokens)],
    "Vocabulary Size": [len(set(hindi_tokens)), len(set(english_tokens))],
    "Average Sentence Length": [
        parallel_df["Hindi Length"].mean(),
        parallel_df["English Length"].mean()
    ]
})

statistics
```

```python
code_mixed_sentences = [
    "आज मेरा Python assignment complete हो गया!",
    "कल college में presentation है।",
    "यह movie बहुत interesting है।",
    "मैंने आज नया laptop खरीदा।",
    "Please मुझे यह file send कर दो।",
    "आज gym के बाद coding करूंगा।",
    "मेरा internet आज बहुत slow है।",
    "इस project का deadline tomorrow है।",
    "तुमने मेरा message check किया?",
    "आज class में machine learning पढ़ी।"
]

code_mixed_df = pd.DataFrame({"Original": code_mixed_sentences})
code_mixed_df
```

```python
def normalize_code_mixed(text):
    text = unicodedata.normalize("NFC", text)
    text = re.sub(r"\s+", " ", text)
    text = re.sub(r"([!?.,])\1+", r"\1", text)
    return text.strip().lower()

code_mixed_df["Normalized"] = code_mixed_df["Original"].apply(
    normalize_code_mixed
)

code_mixed_df
```

```python
DEVANAGARI_RE = re.compile(r"[\u0900-\u097F]")
ROMAN_RE = re.compile(r"[A-Za-z]")

def identify_token(token):
    has_indic = bool(DEVANAGARI_RE.search(token))
    has_roman = bool(ROMAN_RE.search(token))

    if has_indic and has_roman:
        return "Mixed"
    elif has_indic:
        return "Indic"
    elif has_roman:
        return "Roman/English"
    return "Other"

for i, sentence in enumerate(code_mixed_df["Normalized"], 1):
    print(f"Sentence {i}: {sentence}")
    for token in tokenize_text(sentence):
        print(f"  {token} -> {identify_token(token)}")
    print("-" * 50)
```
