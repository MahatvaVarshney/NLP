```python
import math
import numpy as np
import pandas as pd

import nltk
from nltk.tokenize import wordpunct_tokenize
from nltk.corpus import stopwords

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Use NLTK for tokenization and stopword removal.
try:
    STOP_WORDS = set(stopwords.words("english"))
except LookupError:
    # Fallback keeps the notebook usable if the NLTK stopwords corpus
    # has not yet been downloaded in the execution environment.
    STOP_WORDS = {
        "a","an","and","are","as","at","be","been","by","for","from",
        "has","have","he","her","his","i","if","in","is","it","its",
        "me","my","of","on","or","our","she","that","the","their",
        "them","there","these","they","this","to","was","we","were",
        "what","when","where","which","who","will","with","you","your"
    }

def preprocess(text):
    text = text.lower()
    tokens = wordpunct_tokenize(text)
    tokens = [t for t in tokens if t.isalpha()]
    return [t for t in tokens if t not in STOP_WORDS]

def clean_text(text):
    return " ".join(preprocess(text))

print("Libraries loaded successfully.")
print("Stopwords available:", len(STOP_WORDS))
```

```python
# Q2(a): 10 short documents from technology, education and sports.
documents = [
    "Artificial intelligence systems learn patterns from data and make predictions.",
    "Machine learning algorithms train on data to predict outcomes and detect patterns.",
    "Deep learning uses neural networks to learn representations from large datasets.",
    "Computer vision uses deep learning to classify images and detect objects.",
    "Natural language processing analyzes text and helps computers understand language.",
    "Students learn programming algorithms and data structures in computer science courses.",
    "Online education platforms provide courses lectures quizzes and programming exercises.",
    "Sports teams use data analysis to study player performance and improve training.",
    "Football coaches analyze player statistics and team performance using data.",
    "Researchers develop new machine learning methods for language vision and data analysis."
]

processed = [clean_text(doc) for doc in documents]

for i, doc in enumerate(processed, 1):
    print(f"D{i}: {doc}")
```

```python
# Q2(b/c): Generate TF-IDF document vectors for the VSM.
vectorizer = TfidfVectorizer()
matrix = vectorizer.fit_transform(processed)
features = vectorizer.get_feature_names_out()

document_term_matrix = pd.DataFrame(
    matrix.toarray(),
    index=[f"D{i}" for i in range(1, len(documents) + 1)],
    columns=features
)

print("Document-Term Matrix")
display(document_term_matrix.round(4))
print("Dimensionality (documents × terms):", matrix.shape)
```

```python
# Q2(c): Show numerical vector representations.
for i in range(min(3, matrix.shape[0])):
    print(f"D{i+1} vector:")
    print(np.round(matrix[i].toarray().flatten(), 4))
    print()
```

```python
# Q2(e/f): Calculate cosine similarity for at least three document pairs.
pairs = [(1, 2), (3, 4), (8, 9), (1, 5), (6, 7)]

results = []
for a, b in pairs:
    score = cosine_similarity(matrix[a-1], matrix[b-1])[0, 0]
    results.append([f"D{a} - D{b}", score])

similarity_table = pd.DataFrame(results, columns=["Document Pair", "Cosine Similarity"])
display(similarity_table.round(4))
```

```python
# Q2(g): Find the most similar pair among all documents.
all_similarity = cosine_similarity(matrix)
np.fill_diagonal(all_similarity, -1)

i, j = np.unravel_index(np.argmax(all_similarity), all_similarity.shape)

print(f"Most similar pair: D{i+1} and D{j+1}")
print(f"Cosine similarity: {all_similarity[i, j]:.4f}")
```

### Q2(h): How does VSM convert text into numerical vectors?

A Vector Space Model:
1. Builds a vocabulary from the documents.
2. Assigns one vector dimension to each vocabulary term.
3. Calculates a numerical weight such as TF-IDF for each term.
4. Represents each document by one fixed-length numerical vector.
5. Uses vector operations such as cosine similarity to compare documents.

A higher cosine similarity indicates greater similarity in the direction of the document vectors.


### Observation

Documents discussing related topics tend to share important vocabulary and therefore can have higher cosine similarity. The exact similarity depends on the corpus, preprocessing and TF-IDF weighting.
