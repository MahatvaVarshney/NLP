```python
import math
import numpy as np
import pandas as pd

import nltk
from nltk.tokenize import wordpunct_tokenize
from nltk.corpus import stopwords

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Use NLTK for tokenization and stopword removal.
try:
    STOP_WORDS = set(stopwords.words("english"))
except LookupError:
    # Fallback keeps the notebook usable if the NLTK stopwords corpus
    # has not yet been downloaded in the execution environment.
    STOP_WORDS = {
        "a","an","and","are","as","at","be","been","by","for","from",
        "has","have","he","her","his","i","if","in","is","it","its",
        "me","my","of","on","or","our","she","that","the","their",
        "them","there","these","they","this","to","was","we","were",
        "what","when","where","which","who","will","with","you","your"
    }

def preprocess(text):
    text = text.lower()
    tokens = wordpunct_tokenize(text)
    tokens = [t for t in tokens if t.isalpha()]
    return [t for t in tokens if t not in STOP_WORDS]

def clean_text(text):
    return " ".join(preprocess(text))

print("Libraries loaded successfully.")
print("Stopwords available:", len(STOP_WORDS))
```

```python
# Q4(a): 12 sentences from three topics.
sentences = [
    "Machine learning models learn patterns from training data.",
    "Deep learning networks classify images and recognize objects.",
    "Natural language processing models analyze text and language.",
    "Artificial intelligence systems make predictions from data.",
    "Computer vision detects objects in images and videos.",
    "Data scientists build predictive models using algorithms.",
    "Students attend online classes and complete programming exercises.",
    "Teachers design courses and evaluate student learning.",
    "Online education provides lectures quizzes and assignments.",
    "Students practice programming and study computer science concepts.",
    "Sports teams analyze player statistics to improve performance.",
    "Coaches use data to plan training and evaluate players."
]

topics = [
    "AI", "AI", "AI", "AI", "AI", "AI",
    "Education", "Education", "Education", "Education",
    "Sports", "Sports"
]

for i, (s, topic) in enumerate(zip(sentences, topics), 1):
    print(f"S{i} [{topic}]: {clean_text(s)}")
```

```python
# Q4(b): TF-IDF sentence representation.
vectorizer = TfidfVectorizer()
sentence_matrix = vectorizer.fit_transform(sentences)
features = vectorizer.get_feature_names_out()

sentence_vectors = pd.DataFrame(
    sentence_matrix.toarray(),
    index=[f"S{i}" for i in range(1, len(sentences) + 1)],
    columns=features
)

print("Sentence-vector matrix dimensions:", sentence_matrix.shape)
display(sentence_vectors.round(4))
```

```python
# Q4(c): Show the non-zero components of each sentence vector.
for i in range(sentence_matrix.shape[0]):
    row = sentence_matrix.getrow(i)
    entries = [
        f"{features[col]}={value:.3f}"
        for col, value in zip(row.indices, row.data)
    ]
    print(f"S{i+1}: " + ", ".join(entries))
```

```python
# Q4(d): Calculate cosine similarity for at least five sentence pairs.
pairs = [
    (1, 4),
    (2, 5),
    (1, 6),
    (7, 9),
    (8, 10),
    (11, 12),
    (1, 9)
]

results = []
for a, b in pairs:
    score = cosine_similarity(sentence_matrix[a-1], sentence_matrix[b-1])[0, 0]
    results.append([
        f"S{a} - S{b}",
        f"{topics[a-1]} / {topics[b-1]}",
        score
    ])

similarity_table = pd.DataFrame(
    results,
    columns=["Sentence Pair", "Topics", "Cosine Similarity"]
)

display(similarity_table.round(4))
```

```python
# Q4(e): Identify the most similar sentence pairs in the complete corpus.
all_similarity = cosine_similarity(sentence_matrix)
np.fill_diagonal(all_similarity, -1)

pairs_all = []
for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        pairs_all.append([
            f"S{i+1}", f"S{j+1}",
            topics[i], topics[j],
            all_similarity[i, j]
        ])

top_pairs = pd.DataFrame(
    pairs_all,
    columns=["Sentence 1", "Sentence 2", "Topic 1", "Topic 2", "Cosine Similarity"]
).sort_values("Cosine Similarity", ascending=False)

display(top_pairs.head(5).round(4))
```

```python
# Q4(f): Create document vectors by averaging sentence vectors.
# Document 1: AI sentences
# Document 2: Education sentences
# Document 3: Sports sentences

groups = {
    "Document 1 - AI": [0, 1, 2, 3, 4, 5],
    "Document 2 - Education": [6, 7, 8, 9],
    "Document 3 - Sports": [10, 11]
}

document_vectors = {}

for name, indices in groups.items():
    # Mean pooling combines multiple sentence vectors into one fixed-length vector.
    document_vectors[name] = np.asarray(
        sentence_matrix[indices].mean(axis=0)
    ).flatten()

document_matrix = pd.DataFrame(document_vectors).T

print("Document-to-vector dimensions:", document_matrix.shape)
display(document_matrix.round(4))
```

```python
# Compare the generated document vectors.
document_similarity = cosine_similarity(document_matrix.values)

document_similarity_df = pd.DataFrame(
    document_similarity,
    index=document_matrix.index,
    columns=document_matrix.index
)

display(document_similarity_df.round(4))
```

### Q4(g): Comparison of representations

| Representation | Unit | Typical dimension in this notebook | Main information |
|---|---|---:|---|
| Word co-occurrence vector | Word | Corpus vocabulary size | Context words around a word |
| Sentence TF-IDF vector | Sentence | Sentence vocabulary size | Importance of terms in the sentence |
| Document vector | Document | Same feature space as sentence vectors | Aggregated sentence information |

The document vectors here are obtained by mean-pooling sentence vectors, so their dimension remains the same as the sentence-vector dimension.


### Q4(h): Challenges of fixed-length sentence/document vectors

- Word order is not fully represented by simple TF-IDF.
- Synonyms may appear as different dimensions.
- Polysemous words can have different meanings in different contexts.
- Long documents may lose important detail when many vectors are averaged.
- Unseen vocabulary may not have a representation.
- Simple TF-IDF mainly captures term importance rather than deep semantic relationships.

Therefore, fixed-length vectors are useful for numerical comparison, but the quality of the representation depends on the chosen representation technique.
