```python
# Import required libraries

import nltk
import pandas as pd
import numpy as np

from collections import Counter
from nltk.corpus import brown
from nltk.util import ngrams

print("Libraries imported successfully.")
```

```python
# Download the Brown Corpus

nltk.download('brown')

print("Brown Corpus downloaded successfully.")
```

```python
# Load words from the Brown Corpus news category

news_words = brown.words(categories='news')

print("Number of raw tokens in Brown news category:", len(news_words))
print("First 20 raw tokens:")
print(news_words[:20])
```

```python
# Basic preprocessing:
# 1. Convert tokens to lowercase
# 2. Keep alphabetic tokens only

cleaned_words = [
    word.lower()
    for word in news_words
    if word.isalpha()
]

# Select at least 5,000 tokens
tokens = cleaned_words[:5000]

print("Number of cleaned tokens selected:", len(tokens))
print("First 30 preprocessed tokens:")
print(tokens[:30])
```

```python
# Calculate total number of tokens and vocabulary size

total_tokens = len(tokens)
vocabulary = set(tokens)
vocabulary_size = len(vocabulary)

print("Total Tokens:", total_tokens)
print("Vocabulary Size:", vocabulary_size)
```

```python
# Generate unigram, bigram, and trigram sequences

unigrams = list(ngrams(tokens, 1))
bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))

print("Number of unigrams:", len(unigrams))
print("Number of bigrams:", len(bigrams))
print("Number of trigrams:", len(trigrams))

print("\nSample unigrams:")
print(unigrams[:10])

print("\nSample bigrams:")
print(bigrams[:10])

print("\nSample trigrams:")
print(trigrams[:10])
```

```python
# Calculate frequency distributions

unigram_freq = Counter(unigrams)
bigram_freq = Counter(bigrams)
trigram_freq = Counter(trigrams)

print("Frequency distributions created successfully.")
```

```python
# Display the 10 most frequent unigrams

top_10_unigrams = unigram_freq.most_common(10)

unigram_df = pd.DataFrame(
    top_10_unigrams,
    columns=["Unigram", "Frequency"]
)

unigram_df
```

```python
# Display the 10 most frequent bigrams

top_10_bigrams = bigram_freq.most_common(10)

bigram_df = pd.DataFrame(
    [
        (" ".join(bigram), frequency)
        for bigram, frequency in top_10_bigrams
    ],
    columns=["Bigram", "Frequency"]
)

bigram_df
```

```python
# Display the 10 most frequent trigrams

top_10_trigrams = trigram_freq.most_common(10)

trigram_df = pd.DataFrame(
    [
        (" ".join(trigram), frequency)
        for trigram, frequency in top_10_trigrams
    ],
    columns=["Trigram", "Frequency"]
)

trigram_df
```

### Q1(f) Role of a Language Corpus in Statistical Language Analysis

A language corpus is a collection of text that provides real examples of
language usage. It allows statistical NLP methods to identify patterns
in how words occur individually and in sequences.

In this experiment, the Brown Corpus news category provides a realistic
collection of English text. Token frequencies show which words occur
most often, while bigram and trigram frequencies show relationships
between consecutive words.

N-gram statistics can be used in applications such as language modeling,
text prediction, speech recognition, spelling correction, and other NLP
tasks. Therefore, a corpus provides the empirical data required to
estimate statistical properties of natural language.
