```python


import nltk
import random
from nltk.corpus import brown
from nltk.tag import HiddenMarkovModelTagger

print("Libraries imported successfully.")
```

```python
# Download the required Brown Corpus and universal tagset resources.

nltk.download('brown')
nltk.download('universal_tagset')
nltk.download('punkt')
nltk.download('punkt_tab')

print("NLTK resources are ready.")
```

```python
# Load Brown Corpus sentences with Universal POS tags.

tagged_sentences = brown.tagged_sents(
    categories='news',
    tagset='universal'
)

print("Total tagged sentences:", len(tagged_sentences))
print("Example tagged sentence:")
print(tagged_sentences[0])
```

```python
# Remove empty sentences and prepare the data.

tagged_sentences = [
    sentence for sentence in tagged_sentences
    if sentence
]

# Use a fixed seed so the train/test split is reproducible.
random.seed(42)
random.shuffle(tagged_sentences)

split_index = int(0.90 * len(tagged_sentences))

train_data = tagged_sentences[:split_index]
test_data = tagged_sentences[split_index:]

print("Training sentences:", len(train_data))
print("Testing sentences:", len(test_data))
```

```python
# Train an HMM-based POS tagger.

hmm_tagger = HiddenMarkovModelTagger.train(train_data)

print("HMM POS tagger trained successfully.")
```

```python
# Evaluate the HMM tagger on the held-out test set.

accuracy = hmm_tagger.accuracy(test_data)

print(f"Test Accuracy: {accuracy:.4f}")
print(f"Test Accuracy (%): {accuracy * 100:.2f}%")
```

```python
# Predict POS tags for at least five unseen sentences.

unseen_sentences = [
    "The student reads a book",
    "The researchers developed a new model",
    "She walked to the university",
    "The computer generated useful results",
    "Scientists study the effects of climate change"
]

print("Predicted POS tags:\n")

for i, sentence in enumerate(unseen_sentences, start=1):
    tokens = nltk.word_tokenize(sentence)
    prediction = hmm_tagger.tag(tokens)

    print(f"Sentence {i}: {sentence}")
    print("Predicted tags:", prediction)
    print("-" * 70)
```

```python
# Display one additional example showing the observed words and predicted hidden states.

example_sentence = nltk.word_tokenize("The model predicts language patterns")
example_prediction = hmm_tagger.tag(example_sentence)

print("Observed words:", example_sentence)
print("Predicted POS tags:", example_prediction)
```
